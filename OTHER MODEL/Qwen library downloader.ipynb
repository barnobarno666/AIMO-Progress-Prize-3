```python
import sys

from pathlib import Path



WHEEL_ROOT = Path("AA")

WHEEL_DIR = WHEEL_ROOT #/ "wheels"

WHEEL_DIR.mkdir(parents=True, exist_ok=True)



print("Python:", sys.version)

print("Saving wheels to:", WHEEL_DIR)



# Upgrade packaging tools first.

!{sys.executable} -m pip install -U pip setuptools wheel build



# Build/download wheels for Qwen Agent extras.

!{sys.executable} -m pip wheel --wheel-dir "{WHEEL_DIR}" --prefer-binary "qwen-agent[gui,rag,code_interpreter,mcp]"



# Build/download wheels for SGLang and its extras.

!{sys.executable} -m pip wheel --wheel-dir "{WHEEL_DIR}" --prefer-binary "sglang[all]==0.5.9"



print("Finished building/downloading wheels.")

!ls -lh "{WHEEL_DIR}" | head -n 50
```
