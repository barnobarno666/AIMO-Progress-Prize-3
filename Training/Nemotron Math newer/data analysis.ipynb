```python
import pandas as pd
import ast

def process_conversation(messages):
    # Safely evaluate if the data is stored as a string
    if isinstance(messages, str):
        try:
            messages = ast.literal_eval(messages)
        except (ValueError, SyntaxError):
            return None
            
    # Condition: Drop row if no 'assistant' role exists
    if not any(msg.get('role') == 'assistant' for msg in messages):
        return None

    new_messages = []
    thinking_buffer = []

    for msg in messages:
        role = msg.get('role')
        content = msg.get('content', '')

        if role == 'tool':
            # Accumulate multiple tool calls
            thinking_buffer.append(content)
        elif role == 'assistant':
            # Flush accumulated tool contents into a single 'thinking' block 
            # right before the assistant turn
            if thinking_buffer:
                new_messages.append({
                    'role': 'thinking',
                    'content': '\n\n'.join(thinking_buffer)
                })
                thinking_buffer = []  # Clear buffer for multi-turn robustness
            
            new_messages.append(msg)
        else:
            # System and user messages pass through as-is
            new_messages.append(msg)

    return new_messages
df= df_reloaded.copy()
# 1. Apply the transformation
df_reloaded['messages'] = df_reloaded['messages'].apply(process_conversation)

# 2. Drop rows that returned None (missing assistant or parse errors)
df_reloaded = df_reloaded.dropna(subset=['messages']).reset_index(drop=True)

# Verify the transformation on the index you mentioned
print(df_reloaded['messages'].iloc[100])
```

```python
def convert_to_reference_format(messages):
    if isinstance(messages, str):
        import ast
        try:
            messages = ast.literal_eval(messages)
        except (ValueError, SyntaxError):
            return None
            
    if not isinstance(messages, list):
        return messages

    new_messages = []
    thinking_content = None
    
    for msg in messages:
        role = msg.get('role')
        content = msg.get('content', '')
        
        # In reference format, system messages are omitted.
        if role == 'user':
            new_messages.append({
                'role': 'user',
                'content': content,
                'thinking': None
            })
        elif role == 'thinking':
            thinking_content = content
        elif role == 'assistant':
            new_messages.append({
                'role': 'assistant',
                'content': content,
                'thinking': thinking_content
            })
            thinking_content = None # Reset for next turns
            
    return new_messages

# Apply the conversion to the dataframe
df_reloaded['messages'] = df_reloaded['messages'].apply(convert_to_reference_format)

# Select the same index to verify the results
print(df_reloaded['messages'].iloc[1000])
```
